```python
# Artificial Neural Network using Backpropagation

import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

# XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])

# Create ANN model
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='logistic',
    solver='lbfgs',
    max_iter=2000,
    random_state=1
)

# Train the neural network
model.fit(X, y)

# Test the model
predictions = model.predict(X)

# Display results
print("Input Data:")
print(X)

print("\nActual Output:")
print(y)

print("\nPredicted Output:")
print(predictions)

# Calculate accuracy
accuracy = accuracy_score(y, predictions)
print("\nAccuracy:", accuracy * 100, "%")

# Test a new sample
new_sample = np.array([[0, 1]])
result = model.predict(new_sample)

print("\nNew Input:", new_sample)
print("Predicted Output:", result)
```
