```python
# Credit Score Classification

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Create dataset
data = {
    'Income': [25000, 30000, 40000, 50000, 60000,
               70000, 80000, 90000, 100000, 120000],
    'Debt': [20000, 25000, 30000, 35000, 30000,
             25000, 20000, 15000, 10000, 5000],
    'Payment_History': [1, 0, 1, 1, 1, 1, 1, 1, 1, 1],
    'Credit_Score': ['Low', 'Low', 'Low', 'Medium', 'Medium',
                     'Medium', 'High', 'High', 'High', 'High']
}

df = pd.DataFrame(data)

# Features and target
X = df[['Income', 'Debt', 'Payment_History']]
y = df['Credit_Score']

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create and train model
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Display results
print("Actual Credit Scores:", list(y_test))
print("Predicted Credit Scores:", list(y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred) * 100, "%")

# Predict a new customer's credit score
new_customer = pd.DataFrame({
    'Income': [75000],
    'Debt': [18000],
    'Payment_History': [1]
})

prediction = model.predict(new_customer)

print("New Customer Credit Score:", prediction[0])
```
